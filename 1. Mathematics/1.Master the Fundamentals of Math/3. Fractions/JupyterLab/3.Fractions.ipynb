## Section 1: Equivalent Fractions & Reducing

### 1. Mathematical Theory
The foundational logic of fraction manipulation rests on the **Identity Scaling Rule**. Multiplying or dividing both the numerator ($a$) and the denominator ($b$) of a fraction by the exact same non-zero integer ($c$) alters its structural representation but preserves its precise numeric position on a number line.

* **Scaling Up (Equivalence):** $\frac{a}{b} = \frac{a \times c}{b \times c}$
* **Scaling Down (Reducing):** $\frac{a}{b} = \frac{a \div c}{b \div c}$
* **Irreducible Form:** A fraction is in its lowest terms when $\gcd(a, b) = 1$.

### 2. Professional Bridge
In software engineering and data pipelines, this scaling logic directly underpins **Data Normalization** and **Feature Scaling**. When processing metrics from heterogeneous telemetry sources, raw integers must be transformed into standardized fractional baselines (such as unified percentages or scaled weights between $[0, 1]$) without altering the true relative proportions or introducing floating-point precision loss.

### 3. Programmatic Implementation

```python
def compute_gcd(x, y):
    """Computes the Greatest Common Divisor using the Euclidean Algorithm."""
    while y:
        x, y = y, x % y
    return abs(x)

def reduce_fraction(numerator, denominator):
    """Reduces a fraction to its lowest terms using native logic."""
    if denominator == 0:
        raise ValueError("Denominator cannot be zero.")
    
    gcd = compute_gcd(numerator, denominator)
    return numerator // gcd, denominator // gcd

# Verification using the specific trace from Page 1
raw_num, raw_den = 80, 100
reduced_num, reduced_den = reduce_fraction(raw_num, raw_den)
print(f"Native Reduction: {raw_num}/{raw_den} -> {reduced_num}/{reduced_den}")

# Cross-Cancellation String Verification: (56/27) * (14/21) * (15/22)
# Product of numerators: 56 * 14 * 15 = 11760
# Product of denominators: 27 * 21 * 22 = 12474
prod_num = 56 * 14 * 15
prod_den = 27 * 21 * 22
final_num, final_den = reduce_fraction(prod_num, prod_den)
print(f"Native Cross-Cancellation Output: (56/27) * (14/21) * (15/22) = {prod_num}/{12474} -> {final_num}/{final_den}")

# Standard Library Audit
from fractions import Fraction
lib_fraction = Fraction(raw_num, raw_den)
lib_chain = Fraction(56, 27) * Fraction(14, 21) * Fraction(15, 22)

assert (reduced_num, reduced_den) == (lib_fraction.numerator, lib_fraction.denominator), "Mismatch in basic reduction"
assert (final_num, final_den) == (lib_chain.numerator, lib_chain.denominator), "Mismatch in chain cross-cancellation"
print("System Audit: SUCCESS. Native outputs perfectly match standard library assertions.")
```

## Section 2: Division of Zero & Reciprocals

### 1. Mathematical Theory
Division is algebraically defined as multiplication by a multiplicative inverse (reciprocal). This definition creates strict boundary conditions when interacting with zero.

* **Zero Division:** $\frac{0}{a} = 0$ (where $a \neq 0$). Splitting zero assets across $a$ nodes yields $0$ assets per node.
* **Undefined Boundary:** $\frac{a}{0}$ is strictly undefined. No number multiplied by $0$ can restore $a$.
* **Indeterminate Form:** $\frac{0}{0}$ is indeterminate because any real number could satisfies the equation $0 \times x = 0$, making a unique solution impossible.
* **Reciprocal Rule:** The reciprocal of $\frac{a}{b}$ is $\frac{b}{a}$, such that $\frac{a}{b} \times \frac{b}{a} = 1$. The number $0$ has no reciprocal because $\frac{1}{0}$ breaks system boundaries.

### 2. Professional Bridge
In computing environments, violating these boundary conditions triggers immediate system faults. Division by zero causes runtime `ZeroDivisionError` exceptions or results in `NaN` (Not a Number) / `Inf` values in floating-point operations. In distributed systems, failing to catch an unhandled division by zero in an analytics pipeline can halt a service worker thread, creating system crashes or data ingestion bottlenecks.

### 3. Programmatic Implementation

```python
def handle_division(numerator, denominator):
    """Evaluates fractional division and intercepts systemic boundaries."""
    if numerator == 0 and denominator == 0:
        return "Indeterminate (Undefined)"
    if denominator == 0:
        return "Undefined (System Fault)"
    return f"{numerator / denominator}"

def get_reciprocal(numerator, denominator):
    """Inverts a fraction to find its reciprocal, checking boundary safety."""
    if numerator == 0:
        return "Undefined (No reciprocal for zero)"
    
    # Preserve the sign placement cleanly in the string output
    recip_num = denominator
    recip_den = numerator
    return f"{recip_num}/{recip_den}"

# Executing Boundary Logic Checks
print(f"Zero Division check (0 / 5) = {handle_division(0, 5)}")
print(f"Undefined Boundary check (5 / 0) = {handle_division(5, 0)}")
print(f"Indeterminate Boundary check (0 / 0) = {handle_division(0, 0)}")
print(f"Reciprocal of -11/4 = {get_reciprocal(-11, 4)}")
print(f"Reciprocal of 0/1 = {get_reciprocal(0, 1)}")

# Standard Library Audit
from fractions import Fraction
try:
    val = Fraction(5, 0)
except ZeroDivisionError as e:
    print(f"Library Audit Exception Caught: 5/0 -> {e}")
```

## Section 3: Signs, Multiplying & Dividing Fractions

### 1. Mathematical Theory
Multiplication and division alter the arithmetic state across independent numeric vectors, while sign transformation preserves scalar identity using logical sign matrices.

* **Multiplication Rule:** $\frac{a}{b} \times \frac{c}{d} = \frac{a \times c}{b \times d}$
* **Division Rule (Keep-Change-Flip):** $\frac{a}{b} \div \frac{c}{d} = \frac{a}{b} \times \frac{d}{c} = \frac{a \times d}{b \times c}$
* **Sign Shifting Matrix:** A fraction contains three sign distinct zones: $\text{Global Front } \pm \left( \frac{\text{Numerator } \pm}{\text{Denominator } \pm} \right)$. Changing exactly two zones simultaneously neutralizes the operational shift, leaving the absolute scalar value invariant:
    $$-\left(\frac{a}{b}\right) = \frac{-a}{b} = \frac{a}{-b}$$

### 2. Professional Bridge
The Sign Shifting Rule maps directly to **Complement Arithmetic** and **Bitwise Negation Overrides** in low-level compiler optimization. Changing sign states across denominators maps directly to string and parsing transformations within compilers optimizing rational algebraic expressions during Abstract Syntax Tree (AST) evaluation passes.

### 3. Programmatic Implementation

```python
def multiply_fractions(f1_num, f1_den, f2_num, f2_den):
    """Multiplies two fractions straight across and reduces the output."""
    raw_num = f1_num * f2_num
    raw_den = f1_den * f2_den
    return reduce_fraction(raw_num, raw_den)

def divide_fractions(f1_num, f1_den, f2_num, f2_den):
    """Divides fractions by executing Keep-Change-Flip logic."""
    if f2_num == 0:
        raise ZeroDivisionError("Cannot divide by a fraction equal to zero.")
    # Keep-Change-Flip: Multiply by reciprocal of second fraction
    return multiply_fractions(f1_num, f1_den, f2_den, f2_num)

# Evaluating Page 3 Reference Traces
m_num, m_den = multiply_fractions(2, 3, 5, 7)
d_num, d_den = divide_fractions(5, 8, 2, 3)

print(f"Native Multiplication (2/3 * 5/7): {m_num}/{m_den}")
print(f"Native Division (5/8 / 2/3): {d_num}/{d_den}")

# Standard Library Validation
lib_m = Fraction(2, 3) * Fraction(5, 7)
lib_d = Fraction(5, 8) / Fraction(2, 3)

assert (m_num, m_den) == (lib_m.numerator, lib_m.denominator)
assert (d_num, d_den) == (lib_d.numerator, lib_d.denominator)
print("System Audit: SUCCESS. Sign/Operational outputs match standard library assertions.")
```

## Section 4: Adding & Subtracting Fractions

### 1. Mathematical Theory
Addition and subtraction operations cannot be executed directly across disparate denominators because the fractional terms represent completely different scalar partitions. 

* **The Constraint:** $\frac{a}{b} \pm \frac{c}{d} \neq \frac{a \pm c}{b \pm d}$
* **The Operational Pipeline:**
    1. Determine the Least Common Denominator (LCD), which is the Least Common Multiple (LCM) of the denominators: $\text{LCD} = \text{lcm}(b, d)$.
    2. Normalize each numerator by multiplying it by the scaling factor needed to transition its original base denominator to the uniform LCD.
    3. Aggregate or subtract the synchronized numerators over the baseline LCD: $\frac{a \cdot ( \text{LCD}/b ) \pm c \cdot ( \text{LCD}/d )}{\text{LCD}}$.

### 2. Professional Bridge
In distributed database architectures and scheduling layers (like cron clustering or event sync tasks), combining metrics with distinct periodic cycles requires synchronization over a common baseline. Calculating the LCD maps directly to resolving race conditions, optimizing thread pooling frequencies, and aligning mismatched array offsets within high-frequency scheduling matrices.

### 3. Programmatic Implementation

```python
def compute_lcm(x, y):
    """Computes the Least Common Multiple using GCD relational logic."""
    if x == 0 or y == 0:
        return 0
    return abs(x * y) // compute_gcd(x, y)

def add_fractions(n1, d1, n2, d2):
    """Synchronizes denominators over LCD to perform fractional addition."""
    lcd = compute_lcm(d1, d2)
    
    # Scale numerators up to meet the uniform LCD matrix baseline
    scaled_n1 = n1 * (lcd // d1)
    scaled_n2 = n2 * (lcd // d2)
    
    # Combine over the unified baseline
    final_num = scaled_n1 + scaled_n2
    return reduce_fraction(final_num, lcd)

# Evaluating the complex notebook problem from Page 4: 13/40 + 23/65
n1, d1 = 13, 40
n2, d2 = 23, 65

calculated_lcd = compute_lcm(d1, d2)
res_num, res_den = add_fractions(n1, d1, n2, d2)

# Printing the input, step-by-step scaling annotations, and final result cleanly
print("Native Additive Combination Process:")
print(f"Input Problem: {n1}/{d1} + {n2}/{d2}")
print(f"Step 1 (Scale numerators based on LCD {calculated_lcd}): ({n1}*{calculated_lcd//d1})/{calculated_lcd} + ({n2}*{calculated_lcd//d2})/{calculated_lcd}")
print(f"Step 2 (Evaluate scalar products): {n1*(calculated_lcd//d1)}/{calculated_lcd} + {n2*(calculated_lcd//d2)}/{calculated_lcd}")
print(f"Step 3 (Merge numerators over unified baseline): ({n1*(calculated_lcd//d1)} + {n2*(calculated_lcd//d2)})/{calculated_lcd}")
print(f"Step 4 (Reduce to lowest terms): {res_num}/{res_den}")

# Standard Library Audit
from fractions import Fraction
lib_sum = Fraction(n1, d1) + Fraction(n2, d2)

assert calculated_lcd == 520, "LCD tracking logic failure"
assert (res_num, res_den) == (lib_sum.numerator, lib_sum.denominator), "Additive computation mismatch"
print("\nSystem Audit: SUCCESS. Additive normalization matches standard library assertions.")
```
