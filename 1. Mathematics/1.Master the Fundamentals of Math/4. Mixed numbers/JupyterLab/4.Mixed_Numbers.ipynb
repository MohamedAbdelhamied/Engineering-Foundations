## 1. Mixed Numbers & Improper Fractions

### 1.1 Definitions
* **Mixed Number**: A combination of a whole number and a proper fraction.
    * *Form*: $a \frac{b}{c}$
* **Improper Fraction**: A fraction where the numerator is greater than or equal to the denominator.
    * *Form*: $\frac{n}{d}$

### 1.2 Conversion: Mixed Number -> Improper Fraction
* **The Logic**: Represent the whole number as a fraction with the same denominator, then add the existing fraction.
* **The Formula**: New Numerator = (Denominator x Whole Number) + Numerator
* *Example*: $3 \frac{1}{2}$
    * $(2 \times 3) + 1 = 7$
    * Result: $\frac{7}{2}$

### 1.3 Conversion: Improper Fraction -> Mixed Number
* **The Logic**: Divide the numerator by the denominator.
    * **Quotient** = New Whole Number
    * **Remainder** = New Numerator
    * **Divisor** = Original Denominator (stays the same)
* *Example*: $\frac{7}{2}$
    * $7 \div 2 = 3$ with remainder $1$.
    * Result: $3 \frac{1}{2}$

### Programmatic Implementation

```python
import math

def mixed_to_improper(whole, num, den):
    """
    Converts a mixed number to an improper fraction.
    """
    print(f"Input Mixed Number: {whole} {num}/{den}")
    
    # Calculation: (denominator * whole) + numerator
    step_calculation = (den * whole) + num
    print(f"Step 1: Multiply denominator by whole and add numerator -> ({den} * {whole}) + {num} = {step_calculation}")
    
    print(f"Final Result: {step_calculation}/{den}")

def improper_to_mixed(num, den):
    """
    Converts an improper fraction to a mixed number using integer division and modulo.
    """
    print(f"Input Improper Fraction: {num}/{den}")
    
    # Step 1: Quotient becomes the whole number
    whole = num // den
    # Step 2: Remainder becomes the new numerator
    remainder = num % den
    print(f"Step 1 & 2: Divide numerator by denominator -> {num} // {den} = {whole} (Quotient), {num} % {den} = {remainder} (Remainder)")
    
    print(f"Final Result: {whole} {remainder}/{den}")

# Executing Examples
mixed_to_improper(3, 1, 2)
print("-" * 40)
improper_to_mixed(7, 2)
```

## 2. Addition & Subtraction of Mixed Numbers

### 2.1 Addition Strategy
* **Step 1**: Add the whole numbers together separately.
* **Step 2**: Add the fractions using the standard fraction addition rule (Find common denominators if necessary).
* **Step 3**: Simplify the resulting fraction to its lowest terms.
* **Step 4**: If the resulting fraction is an improper fraction, convert it to a mixed number and add the whole number component to the sum from Step 1.

#### Example: $2 \frac{1}{4} + 1 \frac{1}{5}$
* $2 + 1 + \frac{1}{4} + \frac{1}{5}$
* $3 + \frac{1}{4} + \frac{1}{5}$ common denominators is 20
* $3 + \frac{5}{20} + \frac{4}{20} => 3 + \frac{9}{20} => 3 \frac{9}{20}$

### Programmatic Implementation

```python
def add_mixed_numbers(w1, n1, d1, w2, n2, d2):
    """
    Function to add two mixed numbers following the handwritten notebook steps.
    """
    print(f"Input Equation: ({w1} + {n1}/{d1}) + ({w2} + {n2}/{d2})")
    
    # Step 1: Add whole numbers together separately
    sum_whole = w1 + w2
    print(f"Step 1: Add whole numbers together separately -> {w1} + {w2} = {sum_whole}")
    
    # Step 2: Add the fractions (Find common denominators if necessary)
    common_den = (d1 * d2) // math.gcd(d1, d2)
    n1_adj = n1 * (common_den // d1)
    n2_adj = n2 * (common_den // d2)
    sum_num = n1_adj + n2_adj
    print(f"Step 2: Find common denominator ({common_den}) and add fractions -> {n1_adj}/{common_den} + {n2_adj}/{common_den} = {sum_num}/{common_den}")
    
    # Step 3: Simplify the resulting fraction to its lowest terms
    gcd_val = math.gcd(sum_num, common_den)
    sim_num = sum_num // gcd_val
    sim_den = common_den // gcd_val
    print(f"Step 3: Simplify fraction to lowest terms -> {sim_num}/{sim_den}")
    
    # Step 4: If improper fraction, convert it to a mixed number and adjust whole number
    if sim_num >= sim_den:
        extra_whole = sim_num // sim_den
        sim_num = sim_num % sim_den
        sum_whole += extra_whole
        print(f"Step 4: Convert improper fraction to mixed number -> Added {extra_whole} to whole number. Remainder: {sim_num}/{sim_den}")
    else:
        print("Step 4: Fraction is proper. No conversion needed.")
        
    print(f"Final Result: {sum_whole} {sim_num}/{sim_den}")
    return sum_whole, sim_num, sim_den

# Executing Example
add_mixed_numbers(2, 1, 4, 1, 1, 5)
```

### 2.2 Subtraction Strategy
* **Step 1**: Subtract the whole numbers.
* **Step 2**: Subtract the fractions.
* **Step 3**: If the fraction being subtracted is larger than the fraction in the minuend, borrow 1 from the whole number of the minuend.
    * *Mechanism*: Convert the borrowed 1 into a fraction equal to $\frac{\text{denominator}}{\text{denominator}}$ and add it to the existing fraction.
* **Step 4**: Simplify the final result.

#### Example: $11 \frac{2}{4} - 1 \frac{4}{6}$
* $11 - 1 - \frac{2}{4} - \frac{4}{6}$
* $10 + \frac{2}{4} - \frac{4}{6}$ common denominators is 12
* $10 + \frac{6}{12} - \frac{8}{12} => 9 - \frac{2}{12} + \frac{12}{12} => 9 + \frac{10}{12} = 9 \frac{10}{12} => 9 \frac{5}{6}$

### Programmatic Implementation

```python
def subtract_mixed_numbers(w1, n1, d1, w2, n2, d2):
    """
    Function to subtract two mixed numbers following the handwritten notebook steps.
    """
    print(f"Input Equation: ({w1} + {n1}/{d1}) - ({w2} + {n2}/{d2})")
    
    # Step 1: Subtract the whole numbers
    rem_whole = w1 - w2
    print(f"Step 1: Subtract whole numbers -> {w1} - {w2} = {rem_whole}")
    
    # Step 2: Convert to common denominator to compare and subtract fractions
    common_den = (d1 * d2) // math.gcd(d1, d2)
    n1_adj = n1 * (common_den // d1)
    n2_adj = n2 * (common_den // d2)
    print(f"Step 2: Common denominator is {common_den} -> {n1_adj}/{common_den} - {n2_adj}/{common_den}")
    
    # Step 3: Check if borrowing is required
    if n1_adj < n2_adj:
        print(f"Step 3: Fraction being subtracted ({n2_adj}/{common_den}) is larger than minuend fraction ({n1_adj}/{common_den}). Borrowing 1...")
        rem_whole -= 1
        n1_adj += common_den
        print(f"Mechanism: Decomposing whole number to add {common_den}/{common_den} to minuend fraction -> New whole: {rem_whole}, New fraction numerator: {n1_adj}")
        
    diff_num = n1_adj - n2_adj
    print(f"Perform subtraction -> Fraction outcome: {diff_num}/{common_den}")
    
    # Step 4: Simplify the final result
    gcd_val = math.gcd(diff_num, common_den)
    sim_num = diff_num // gcd_val
    sim_den = common_den // gcd_val
    
    print(f"Step 4: Simplify the final result -> {sim_num}/{sim_den}")
    print(f"Final Result: {rem_whole} {sim_num}/{sim_den}")

# Executing Example
subtract_mixed_numbers(11, 2, 4, 1, 4, 6)
```

## 3. Multiplication & Division of Mixed Numbers

### 3.1 Multiply Strategy
* **Step 1**: Convert both mixed numbers into improper fractions.
* **Step 2**: Multiply the numerators together and the denominators together (Standard Fraction multiplication).
* **Step 3**: Simplify the resulting fraction to its lowest terms.
* **Step 4**: (optional) Convert the result back into a mixed number.

#### Example: $4 \frac{1}{2} \times 5 \frac{3}{4}$
* $\frac{9}{2} \times \frac{23}{4} = \frac{207}{8}$
* $25 \frac{7}{8}$

### Programmatic Implementation

```python
# *code block start
def multiply_mixed_numbers(w1, n1, d1, w2, n2, d2):
    """
    Function to multiply two mixed numbers by converting to improper fractions.
    """
    print(f"Input Equation: ({w1} {n1}/{d1}) * ({w2} {n2}/{d2})")
    
    # Step 1: Convert both mixed numbers into improper fractions
    imp_num1 = (w1 * d1) + n1
    imp_num2 = (w2 * d2) + n2
    print(f"Step 1: Convert to improper fractions -> {imp_num1}/{d1} * {imp_num2}/{d2}")
    
    # Step 2: Multiply the numerators together and the denominators together
    res_num = imp_num1 * imp_num2
    res_den = d1 * d2
    print(f"Step 2: Multiply numerators and denominators -> ({imp_num1} * {imp_num2})/({d1} * {d2}) = {res_num}/{res_den}")
    
    # Step 3: Simplify the resulting fraction to its lowest terms
    gcd_val = math.gcd(res_num, res_den)
    sim_num = res_num // gcd_val
    sim_den = res_den // gcd_val
    print(f"Step 3: Simplify to lowest terms -> {sim_num}/{sim_den}")
    
    # Step 4: Convert back to a mixed number
    final_whole = sim_num // sim_den
    final_num = sim_num % sim_den
    print(f"Step 4: Convert back to mixed number -> {final_whole} {final_num}/{sim_den}")

# Executing Example
multiply_mixed_numbers(4, 1, 2, 5, 3, 4)
```

### 3.2 Division Strategy
* **Step 1**: Convert both mixed numbers into improper fractions.
* **Step 2**: Apply the "multiply by the reciprocal" rule: Keep the first fraction, Change the division sign to multiplication, and Flip the second fraction (the divisor).
* **Step 3**: Multiply the numerators and denominators (Standard Fraction multiplication).
* **Step 4**: Simplify and (optional) convert back to a mixed number.

#### Example: $4 \frac{1}{2} \div 5 \frac{3}{4}$
* $\frac{9}{2} \div \frac{23}{4} = \frac{9}{2} \times \frac{4}{23}$
* $\frac{36}{46} = \frac{18}{23}$

### Programmatic Implementation

```python
def divide_mixed_numbers(w1, n1, d1, w2, n2, d2):
    """
    Function to divide mixed numbers using the reciprocal strategy.
    """
    print(f"Input Equation: ({w1} {n1}/{d1}) / ({w2} {n2}/{d2})")
    
    # Step 1: Convert both mixed numbers into improper fractions
    imp_num1 = (w1 * d1) + n1
    imp_num2 = (w2 * d2) + n2
    print(f"Step 1: Convert to improper fractions -> {imp_num1}/{d1} / {imp_num2}/{d2}")
    
    # Step 2: Apply the 'multiply by the reciprocal' rule
    print(f"Step 2: Keep, Change, Flip -> {imp_num1}/{d1} * {imp_num2}/{d2} reciprocal -> {imp_num1}/{d1} * {d2}/{imp_num2}")
    
    # Step 3: Multiply numerators and denominators
    res_num = imp_num1 * d2
    res_den = d1 * imp_num2
    print(f"Step 3: Multiply out -> {res_num}/{res_den}")
    
    # Step 4: Simplify the result
    gcd_val = math.gcd(res_num, res_den)
    sim_num = res_num // gcd_val
    sim_den = res_den // gcd_val
    print(f"Step 4: Simplify to lowest terms -> {sim_num}/{sim_den}")
    
    if sim_num >= sim_den:
        final_whole = sim_num // sim_den
        final_num = sim_num % sim_den
        print(f"Resulting Mixed Number: {final_whole} {final_num}/{sim_den}")
        return final_whole, final_num, sim_den
    else:
        print(f"Resulting Proper Fraction: {sim_num}/{sim_den}")

# Executing Example
divide_mixed_numbers(4, 1, 2, 5, 3, 4)
```

## 4. Relationships of Numbers

### 4.1 Conceptual Objective
* Finding a specific fractional point (e.g., $\frac{1}{5}$ of the way) between two numbers (e.g., $\frac{1}{8}$ and $\frac{9}{11}$) on a number line.

### 4.2 Calculation Strategy
* **Step 1**: Find the total distance between the two points by subtracting the smaller from the larger (End - Start).
* **Step 2**: Calculate the proportional distance by multiplying the total distance by the desired fraction (e.g., Total Distance $\times \frac{1}{5}$).
* **Step 3**: Locate the final point by adding the result from Step 2 to your starting point (Start + Proportional Distance).

#### Example: Find $\frac{1}{5}$ of the way from $\frac{1}{8}$ to $\frac{9}{11}$
* **Step 1**: $\frac{9}{11} - \frac{1}{8} \Rightarrow \frac{8}{8}\left(\frac{9}{11}\right) - \frac{11}{11}\left(\frac{1}{8}\right) \Rightarrow \frac{72}{88} - \frac{11}{88} = \frac{61}{88}$
* **Step 2**: $\frac{61}{88} \times \frac{1}{5} = \frac{61}{440}$
* **Step 3**: $\left(\frac{55}{55}\right)\frac{1}{8} + \frac{61}{440} \Rightarrow \frac{55}{440} + \frac{61}{440} \Rightarrow \frac{116}{440} \Rightarrow \frac{58}{220} = \frac{29}{110}$

### Programmatic Implementation

```python
def find_fractional_distance(start_n, start_d, end_n, end_d, target_n, target_d):
    """
    Finds a specific fractional target distance between two fractions on a number line.
    """
    print(f"Goal: Find {target_n}/{target_d} of the way from {start_n}/{start_d} to {end_n}/{end_d}")
    
    # Step 1: Total distance = End - Start
    com_den1 = start_d * end_d
    adj_end_n = end_n * start_d
    adj_start_n = start_n * end_d
    dist_n = adj_end_n - adj_start_n
    dist_d = com_den1
    print(f"Step 1: Distance calculation ({end_n}/{end_d} - {start_n}/{start_d}) -> {adj_end_n}/{com_den1} - {adj_start_n}/{com_den1} = {dist_n}/{dist_d}")
    
    # Step 2: Proportional distance = Total Distance * Desired fraction
    prop_n = dist_n * target_n
    prop_d = dist_d * target_d
    print(f"Step 2: Proportional distance ({dist_n}/{dist_d} * {target_n}/{target_d}) -> {prop_n}/{prop_d}")
    
    # Step 3: Final Point = Start + Proportional Distance
    final_com_den = (start_d * prop_d) // math.gcd(start_d, prop_d)
    scaled_start_n = start_n * (final_com_den // start_d)
    scaled_prop_n = prop_n * (final_com_den // prop_d)
    
    ans_n = scaled_start_n + scaled_prop_n
    ans_d = final_com_den
    print(f"Step 3: Locate final point ({start_n}/{start_d} + {prop_n}/{prop_d}) -> {scaled_start_n}/{final_com_den} + {scaled_prop_n}/{final_com_den} = {ans_n}/{ans_d}")
    
    # Simplify the final answer
    gcd_final = math.gcd(ans_n, ans_d)
    sim_ans_n = ans_n // gcd_final
    sim_ans_d = ans_d // gcd_final
    print(f"Simplify final answer -> {sim_ans_n}/{sim_ans_d}")

# Executing Example
find_fractional_distance(1, 8, 9, 11, 1, 5)
```

## 5. Adding Mixed Measures

### 5.1 Conceptual Objective
* Perform addition across different unit systems (e.g., hours/minutes/seconds or yards/feet/inches) by treating each unit as a distinct column.

### 5.2 Calculation Strategy
* **Step 1**: Arrange the measures in columns, aligning like units (e.g., hours under hours, minutes under minutes).
* **Step 2**: Sum each column independently.
* **Step 3**: Normalize the results starting from the smallest unit. If a column exceeds the conversion threshold (e.g., $\ge 60$ minutes or $\ge 12$ inches), convert the excess into the next larger unit and carry it over.
* **Step 4**: Simplify the final values to their smallest representation.

#### Example 1: Time Systems
* Row 1: 1 hr 1 min
* Row 2: 2 hr 40 min 35 sec
* Row 3: 1 hr 45 min 50 sec
* Column sum: 4 hr 86 min 85 sec
* Normalized: 4 hr 26 min 25 sec

#### Example 2: Distance Systems
* Row 1: 1 yd 1 ft
* Row 2: 1 yd 2 ft 7 in
* Row 3: 2 yd 2 ft 8 in
* Column sum: 4 yd 5 ft 15 in
* Normalized: 4 yd 2 ft 3 in

### Programmatic Implementation

```python
def add_time_measures(time_list):
    """
    Adds a list of time tuples (hrs, mins, secs) mimicking column addition.
    """
    print("Initial Time Rows to add:")
    for t in time_list:
        print(f"  {t[0]} hr  {t[1]} min  {t[2]} sec")
        
    # Step 1 & 2: Sum each column independently
    total_hr = sum(t[0] for t in time_list)
    total_min = sum(t[1] for t in time_list)
    total_sec = sum(t[2] for t in time_list)
    print(f"Step 2: Column sums -> {total_hr} hr, {total_min} min, {total_sec} sec")
    
    # Step 3: Normalize starting from smallest unit (seconds)
    if total_sec >= 60:
        extra_min = total_sec // 60
        total_sec = total_sec % 60
        total_min += extra_min
        print(f"Normalize Seconds: Carry {extra_min} min from excess seconds. Remaining seconds: {total_sec}")
        
    if total_min >= 60:
        extra_hr = total_min // 60
        total_min = total_min % 60
        total_hr += extra_hr
        print(f"Normalize Minutes: Carry {extra_hr} hr from excess minutes. Remaining minutes: {total_min}")
        
    print(f"Final Simplified Representation: {total_hr} hr {total_min} min {total_sec} sec")
    return total_hr, total_min, total_sec

def add_distance_measures(dist_list):
    """
    Adds a list of imperial units (yds, feet, inches) mimicking column addition.
    """
    print("Initial Distance Rows to add:")
    for d in dist_list:
        print(f"  {d[0]} yd  {d[1]} ft  {d[2]} in")
        
    # Step 1 & 2: Sum each column independently
    total_yd = sum(d[0] for d in dist_list)
    total_ft = sum(d[1] for d in dist_list)
    total_in = sum(d[2] for d in dist_list)
    print(f"Step 2: Column sums -> {total_yd} yd, {total_ft} ft, {total_in} in")
    
    # Step 3: Normalize starting from inches (threshold 12)
    if total_in >= 12:
        extra_ft = total_in // 12
        total_in = total_in % 12
        total_ft += extra_ft
        print(f"Normalize Inches: Carry {extra_ft} ft from excess inches. Remaining inches: {total_in}")
        
    # Normalize feet (threshold 3)
    if total_ft >= 3:
        extra_yd = total_ft // 3
        total_ft = total_ft % 3
        total_yd += extra_yd
        print(f"Normalize Feet: Carry {extra_yd} yd from excess feet. Remaining feet: {total_ft}")
        
    print(f"Final Simplified Representation: {total_yd} yd {total_ft} ft {total_in} in")

# Executing Example 1
time_data = [(2, 40, 35), (1, 45, 50)]
add_time_measures(time_data)

print("-" * 40+"\n")

# Executing Example 2
distance_data = [(1, 2, 7), (2, 2, 8)]
add_distance_measures(distance_data)
```
